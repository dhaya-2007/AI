```python
# Step 1: Import libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

# Step 2: Load the IMDB dataset
vocab_size = 10000
max_length = 200

(X_train, y_train), (X_test, y_test) = imdb.load_data(
    num_words=vocab_size
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# Step 3: Pad sequences to a fixed length
X_train = pad_sequences(
    X_train, maxlen=max_length, padding="post", truncating="post"
)
X_test = pad_sequences(
    X_test, maxlen=max_length, padding="post", truncating="post"
)

print("Padded training shape:", X_train.shape)
print("Padded testing shape:", X_test.shape)

# Step 4: Build the LSTM model
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=64),
    LSTM(64),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Step 5: Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=3,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)

# Step 6: Evaluate the model
test_loss, test_accuracy = model.evaluate(
    X_test, y_test, verbose=0
)

print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Accuracy Percentage: {test_accuracy * 100:.2f}%")

# Step 7: Prepare new review sentences
word_index = imdb.get_word_index()

def preprocess_review(review):
    words = review.lower().split()
    sequence = []

    for word in words:
        # IMDB word indices are offset by 3
        index = word_index.get(word, 2) + 3
        if index >= vocab_size:
            index = 2
        sequence.append(index)

    return pad_sequences(
        [sequence], maxlen=max_length,
        padding="post", truncating="post"
    )

def predict_sentiment(review):
    processed = preprocess_review(review)
    probability = float(model.predict(processed, verbose=0)[0][0])

    sentiment = "Positive" if probability >= 0.5 else "Negative"
    print("\nReview:", review)
    print("Predicted sentiment:", sentiment)
    print(f"Positive probability: {probability:.4f}")

# Step 8: Predict two new reviews
predict_sentiment(
    "This movie was wonderful, exciting and beautifully acted"
)

predict_sentiment(
    "The movie was boring, disappointing and a complete waste of time"
)
```

```python
import pandas as pd

# Replace with the actual path to your uploaded file (e.g., 'IMDB Dataset.csv' or 'imdb.csv')
file_path = 'IMDB Dataset.csv'

try:
    # Load the dataset using pandas
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    print(f"Shape of dataset: {df.shape}")
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file at '{file_path}' was not found. Please verify the filename or upload the file to your Google Colab session first.")
except Exception as e:
    print(f"An error occurred while loading the file: {e}")
```

```python
import re
from tensorflow.keras.preprocessing.text import Tokenizer

# 1. Clean the text (remove HTML tags and non-alphabetical characters)
def clean_text(text):
    text = re.sub(r'<br\s*/?>', ' ', text)  # remove HTML break tags
    text = re.sub(r'[^a-zA-Z\s]', '', text) # keep only letters and spaces
    return text.lower()

df['clean_review'] = df['review'].apply(clean_text)

# 2. Convert labels to binary (1 for positive, 0 for negative)
df['label'] = df['sentiment'].apply(lambda x: 1 if x == 'positive' else 0)

# 3. Tokenize the cleaned text
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(df['clean_review'])

sequences = tokenizer.texts_to_sequences(df['clean_review'])

# 4. Pad the sequences
X_local = pad_sequences(sequences, maxlen=max_length, padding='pre', truncating='post')
y_local = df['label'].values

print("Preprocessed features shape:", X_local.shape)
print("Labels shape:", y_local.shape)
print("Sample processed review sequence:\n", X_local[0][:50])
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# Split using train_test_split (80% train, 20% test)
X_train_local, X_test_local, y_train_local, y_test_local = train_test_split(
    X_local, y_local, test_size=0.2, random_state=42
)

# Rebuild the model optimizing it with masking
local_model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=64, mask_zero=True),
    LSTM(64),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

local_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train the model
history_local = local_model.fit(
    X_train_local,
    y_train_local,
    epochs=3,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)

# Evaluate the model
local_loss, local_accuracy = local_model.evaluate(
    X_test_local, y_test_local, verbose=0
)

print(f"\nLocal Model Test Loss: {local_loss:.4f}")
print(f"Local Model Test Accuracy: {local_accuracy * 100:.2f}%")
```
